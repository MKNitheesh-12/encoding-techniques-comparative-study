```python
import pandas as pd
import numpy as np

# Create the dummy dataset
data = {
    "Patient_ID": ["P1", "P2", "P3", "P4", "P5"],
    "Gender": ["M", "F", "F", "M", "F"],
    "AppointmentType": ["Routine", "Emergency", "Routine", "Follow-up", "Emergency"],
    "City": ["New York", "Los Angeles", "Chicago", "New York", "Chicago"]
}
df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
print("\nDataset Info:")
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy of the original dataframe for label encoding
df_label = df.copy()

# Initialize label encoders for each column
le_gender = LabelEncoder()
le_appt = LabelEncoder()
le_city = LabelEncoder()

# Apply label encoding
df_label['Gender'] = le_gender.fit_transform(df_label['Gender'])
df_label['AppointmentType'] = le_appt.fit_transform(df_label['AppointmentType'])
df_label['City'] = le_city.fit_transform(df_label['City'])

print("Dataset after Label Encoding:")
print(df_label)
print("\nMapping of categories to integers:")
print("Gender: M ->", le_gender.transform(['M'])[0], ", F ->", le_gender.transform(['F'])[0])
print("AppointmentType: Routine ->", le_appt.transform(['Routine'])[0],
      ", Emergency ->", le_appt.transform(['Emergency'])[0],
      ", Follow-up ->", le_appt.transform(['Follow-up'])[0])
print("City: New York ->", le_city.transform(['New York'])[0],
      ", Los Angeles ->", le_city.transform(['Los Angeles'])[0],
      ", Chicago ->", le_city.transform(['Chicago'])[0])
```

```python
# Perform one-hot encoding on the categorical columns
df_onehot = pd.get_dummies(df, columns=['Gender', 'AppointmentType', 'City'])

print("Dataset after One-Hot Encoding:")
print(df_onehot)
print("\nShape of one-hot encoded dataset:", df_onehot.shape)
```
